```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("employee_database.db")
```

```python
conn.execute("""
CREATE TABLE employees (
    emp_id INTEGER,
    emp_name VARCHAR(50),
    department_id INTEGER,
    salary INTEGER,
    location VARCHAR(50)
);
""")
```

```python
conn.execute("""
CREATE TABLE departments (
    department_id INTEGER,
    department_name VARCHAR(50)
);
""")
```

```python
#1. Display all employees
query = """
SELECT *
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 2 — Display employee name and salary
query = """
SELECT emp_name, salary
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 3 — Find unique locations
query = """
SELECT DISTINCT location
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 4 — Employees earning more than 50,000
query = """
SELECT *
FROM employees
WHERE salary > 50000;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 5 — Employees from Pune earning more than 50,000
query = """
SELECT *
FROM employees
WHERE location = 'Pune'
AND salary > 50000;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 6 — Employees from Pune or Mumbai
query = """
SELECT *
FROM employees
WHERE location = 'Pune'
OR location = 'Mumbai';
"""

pd.read_sql_query(query, conn)
```

```python
#Question 7 — Sort employees by salary from highest to lowest
query = """
SELECT *
FROM employees
ORDER BY salary DESC;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 8 — Find the highest salary
query = """
SELECT MAX(salary) AS highest_salary
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 9 — Find average salary
query = """
SELECT AVG(salary) AS average_salary
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 10 — Count total employees
query = """
SELECT COUNT(*) AS total_employees
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 11 — Find total salary
query = """
SELECT SUM(salary) AS total_salary
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 12 — Count employees in each department
query = """
SELECT department_id,
       COUNT(*) AS employee_count
FROM employees
GROUP BY department_id;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 13 — Average salary by departmen
query = """
SELECT department_id,
       AVG(salary) AS average_salary
FROM employees
GROUP BY department_id;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 14 — Departments having more than 5 employees
query = """
SELECT department_id,
       COUNT(*) AS employee_count
FROM employees
GROUP BY department_id
HAVING COUNT(*) > 5;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 15 — Departments with average salary above 60,000
query = """
SELECT department_id,
       AVG(salary) AS average_salary
FROM employees
GROUP BY department_id
HAVING AVG(salary) > 60000;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 16 — Employee name with department name
query = """
SELECT e.emp_name,
       d.department_name
FROM employees e
INNER JOIN departments d
ON e.department_id = d.department_id;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 17 — Display all employees including those without a department
query = """
SELECT e.emp_name,
       d.department_name
FROM employees e
LEFT JOIN departments d
ON e.department_id = d.department_id;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 18 — Categorize employees based on salary
query = """
SELECT emp_name,
       salary,
       CASE
           WHEN salary >= 80000 THEN 'High'
           WHEN salary >= 50000 THEN 'Medium'
           ELSE 'Low'
       END AS salary_category
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 19 — Employees from Pune and Mumbai using UNION
query = """
SELECT emp_name
FROM employees
WHERE location = 'Pune'

UNION

SELECT emp_name
FROM employees
WHERE location = 'Mumbai';
"""

pd.read_sql_query(query, conn)
```

```python
#Question 20 — Convert employee names to uppercase
query = """
SELECT UPPER(emp_name) AS employee_name
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 21 — Employees whose name starts with A
query = """
SELECT *
FROM employees
WHERE emp_name LIKE 'A%';
"""

pd.read_sql_query(query, conn)
```

```python
#Question 22 — Display employee names and their length
query = """
SELECT emp_name,
       LENGTH(emp_name) AS name_length
FROM employees;
"""

pd.read_sql_query(query, conn)
```

```python
#Question 23 — Employees earning more than average salary
query = """
SELECT emp_name,
       salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
"""

pd.read_sql_query(query, conn)
```

```python
#Question 24 — Employee(s) with the highest salary
query = """
SELECT emp_id,
       emp_name,
       salary
FROM employees
WHERE salary = (
    SELECT MAX(salary)
    FROM employees
);
"""

pd.read_sql_query(query, conn)
```

```python
#Question 25 — Highest-paid employee in each department
query = """
SELECT e.emp_id,
       e.emp_name,
       e.department_id,
       e.salary
FROM employees e
WHERE e.salary = (
    SELECT MAX(e2.salary)
    FROM employees e2
    WHERE e2.department_id = e.department_id
);
"""

pd.read_sql_query(query, conn)
```
